# Importando librerias

```python
import pandas as pd
import folium as fl
import plotly.express as px
```

# Definiendo la variable del dataset

```python
felinos = pd.read_csv("felinos_filtrado.csv")
```

## 1) ¿Cuantos avistamientos hay por especie? (gráfico barra)

```python
f_s= felinos["species"].value_counts()
px.bar(data_frame=f_s, x=f_s.index, y=f_s.values, title="Cantidad de felinos por especie en grafico de barra.").show()
```

## 2) ¿Cual es el porcentaje de cada especie respecto al total de especies de avistamientos? (gráfico torta)

```python
# uso f_s definido en la anterior actividad
# formato de torta px.pie(data_frame=VARIABLE_DEL_FILTRO, values=VARIABLE_DEL_FILTRO.values, names=VARIABLE_DEL_FILTRO.index,title="titulo").show() 
px.pie(data_frame=f_s, values=f_s.values, names=f_s.index, title="Porcentaje de cada especie de avistajes").show()
```

## 3) ¿Qué provincias tienen mayor cantidad de avistamientos? (gráfico de barra)

```python
f_s_p = felinos["stateProvince"].value_counts()
px.bar(data_frame=f_s_p, x=f_s_p.index, y=f_s_p.values, title="Cantidad de felinos por provincia en grafico de barra.").show()
```
